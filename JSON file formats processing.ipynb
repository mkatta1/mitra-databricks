```sql
/*#Task 4: Read JSON File from Folder
#Reading data from a folder is a unique feature that allows you to append data from multiple tables into a single table. This operation is also dynamic as it adds new records to the data table.
#Step 1: Set Data Schema
#Import all necessary Libraries
*/
from pyspark.sql.types import StructType, StructField, IntegerType, StringType

name_folder_schema = StructType(fields=[StructField("first_name", StringType(), 				True), StructField("last_name", StringType(), True)
])

email_folder_schema = StructType(fields=[StructField("email", StringType(), True)])

nestedjson_folder_schema = StructType(fields=[StructField("id", IntegerType(), False), StructField("name", name_folder_schema), StructField("age", IntegerType(), True), StructField("contact", email_folder_schema)
])
```

```sql
#Step 2: Read Folder

json_folder_nested = spark.read.\
     schema(nestedjson_folder_schema)\
        .json("/mnt/adfdatalakeurbizedge/bronzeunitycatalog/Nest_Folder")

json_folder_nested.show()

Now, let's set the column structure by using the withColumn command.

from pyspark.sql.functions import col, concat, lit
json_folder_nested = json_folder_nested.withColumn("name", concat(col("name.first_name"),lit(" "),col("name.last_name")))

json_folder_nested = json_folder_nested.withColumn("contact",col("contact.email"))
```
